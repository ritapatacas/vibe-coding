```
!pip install requests openai

import requests
from openai import OpenAI

# Read the API key from openai.txt
try:
    with open("openai.txt", "r") as f:
        key = f.read().strip()
    print("API key loaded successfully.")
except FileNotFoundError:
    print("Error: openai.txt not found. Please make sure the file exists.")
    key = None

# Function to get a random quote from the Quotable API
def get_random_quote():
    # Add verify=False to bypass SSL certificate verification
    response = requests.get("https://api.quotable.io/random", verify=False)
    if response.status_code == 200:
        data = response.json()
        return data.get("content")
    else:
        return "Could not fetch a quote at this time."

# Get a random quote
random_quote = get_random_quote()

# Use the random quote as the prompt for ChatGPT
prompt = f"Based on the following quote, write a haiku:\n\n'{random_quote}'"

# Authenticate with your OpenAI API key
if key:
    client = OpenAI(api_key=key)

    print(f"Using the quote as prompt: '{random_quote}'")

    try:
        completion = client.chat.completions.create(
            model="gpt-4o",  # Or another suitable model
            messages=[
                {"role": "system", "content": "You are a helpful assistant that writes haikus."},
                {"role": "user", "content": prompt}
            ]
        )
        # Get the content of the generated message
        generated_text = completion.choices[0].message.content

        # Print the generated text
        print("\nChatGPT Response (Haiku):")
        print(generated_text)

        # Save the generated text to a file (optional)
        with open("generated_haiku_from_quote.txt", "w") as f:
            f.write(generated_text)
        print("\nGenerated text saved to generated_haiku_from_quote.txt")

    except Exception as e:
        print(f"An error occurred with the OpenAI API: {e}")
else:
    print("Cannot proceed without an OpenAI API key.")
```

## <font size="5">2. &nbsp;&nbsp;Quote > Haiku > Image</font></font>

```
!pip install requests openai

import requests
from openai import OpenAI
import base64 # Import base64 for saving the image

# Read the API key from openai.txt
try:
    with open("openai.txt", "r") as f:
        key = f.read().strip()
    print("API key loaded successfully.")
except FileNotFoundError:
    print("Error: openai.txt not found. Please make sure the file exists.")
    key = None

# Function to get a random quote from the Quotable API
def get_random_quote():
    # Add verify=False to bypass SSL certificate verification
    response = requests.get("https://api.quotable.io/random", verify=False)
    if response.status_code == 200:
        data = response.json()
        return data.get("content")
    else:
        return "Could not fetch a quote at this time."

# Get a random quote
random_quote = get_random_quote()

# Use the random quote as the prompt for ChatGPT
prompt = f"Based on the following quote, write a haiku:\n\n'{random_quote}'"

# Authenticate with your OpenAI API key
if key:
    client = OpenAI(api_key=key)

    print(f"Using the quote as prompt: '{random_quote}'")

    try:
        completion = client.chat.completions.create(
            model="gpt-4o",  # Or another suitable model
            messages=[
                {"role": "system", "content": "You are a helpful assistant that writes haikus."},
                {"role": "user", "content": prompt}
            ]
        )
        # Get the content of the generated message
        haiku_text = completion.choices[0].message.content

        # Print the generated text (haiku)
        print("\nChatGPT Response (Haiku):")
        print(haiku_text)

        # Save the generated haiku to a file (optional)
        with open("generated_haiku_from_quote.txt", "w") as f:
            f.write(haiku_text)
        print("\nGenerated haiku saved to generated_haiku_from_quote.txt")

        # Generate an image from the haiku using DALL-E
        print("\nGenerating image from haiku...")
        image_response = client.images.generate(
            model="dall-e-2",  # Or another suitable DALL-E model
            prompt=haiku_text,
            size="1024x1024",  # Or other available sizes
            n=1,  # Number of images to generate
            response_format="b64_json" # To get the image data as base64
        )

        # Get the image data
        image_data = image_response.data[0].b64_json
        print("Image generated successfully.")

        # Save the image data to a file
        with open("generated_image_from_haiku.png", "wb") as f:
            f.write(base64.b64decode(image_data))
        print("Generated image saved to generated_image_from_haiku.png")


    except Exception as e:
        print(f"An error occurred with the OpenAI API: {e}")
else:
    print("Cannot proceed without an OpenAI API key.")

```

## <font size="5">3. &nbsp;&nbsp;Quote > Haiku > Image > Email </font>

> 🔎&nbsp;&nbsp;<font size="3" color="lightsteelblue">Make a program that:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;1. Generates a random **quote**<br>

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;4. **Emails** you the image with the title "Your AI poetry"<br><br><font size="3" color="gray">

```
!pip install requests openai

import requests
from openai import OpenAI
import base64 # Import base64 for saving the image
import smtplib
from email.mime.text import MIMEText
from email.mime.image import MIMEImage
from email.mime.multipart import MIMEMultipart
from google.colab import userdata


# Read the API key from openai.txt
try:
    with open("openai.txt", "r") as f:
        key = f.read().strip()
    print("API key loaded successfully.")
except FileNotFoundError:
    print("Error: openai.txt not found. Please make sure the file exists.")
    key = None

# Function to get a random quote from the Quotable API
def get_random_quote():
    # Add verify=False to bypass SSL certificate verification
    response = requests.get("https://api.quotable.io/random", verify=False)
    if response.status_code == 200:
        data = response.json()
        return data.get("content")
    else:
        return "Could not fetch a quote at this time."

# Get a random quote
random_quote = get_random_quote()

# Use the random quote as the prompt for ChatGPT
prompt = f"Based on the following quote, write a haiku:\n\n'{random_quote}'"

# Authenticate with your OpenAI API key
if key:
    client = OpenAI(api_key=key)

    print(f"Using the quote as prompt: '{random_quote}'")

    try:
        completion = client.chat.completions.create(
            model="gpt-4o",  # Or another suitable model
            messages=[
                {"role": "system", "content": "You are a helpful assistant that writes haikus."},
                {"role": "user", "content": prompt}
            ]
        )
        # Get the content of the generated message
        haiku_text = completion.choices[0].message.content

        # Print the generated text (haiku)
        print("\nChatGPT Response (Haiku):")
        print(haiku_text)

        # Save the generated haiku to a file (optional)
        with open("generated_haiku_from_quote.txt", "w") as f:
            f.write(haiku_text)
        print("\nGenerated haiku saved to generated_haiku_from_quote.txt")

        # Generate an image from the haiku using DALL-E
        print("\nGenerating image from haiku...")
        image_response = client.images.generate(
            model="dall-e-2",  # Or another suitable DALL-E model
            prompt=haiku_text,
            size="1024x1024",  # Or other available sizes
            n=1,  # Number of images to generate
            response_format="b64_json" # To get the image data as base64
        )

        # Get the image data
        image_data = image_response.data[0].b64_json
        print("Image generated successfully.")

        # Save the image data to a file
        image_filename = "generated_image_from_haiku.png"
        with open(image_filename, "wb") as f:
            f.write(base64.b64decode(image_data))
        print(f"Generated image saved to {image_filename}")

        # Email the image
        SENDER_EMAIL = 'devritsa@gmail.com'
        SENDER_PASSWORD = userdata.get('GMAIL_PASSWORD')
        RECEIVER_EMAIL = SENDER_EMAIL  # Sending the email to yourself for this example

        if SENDER_EMAIL and SENDER_PASSWORD:
            msg = MIMEMultipart()
            msg['Subject'] = 'Your AI poetry'
            msg['From'] = SENDER_EMAIL
            msg['To'] = RECEIVER_EMAIL

            # Attach the image
            with open(image_filename, 'rb') as f:
                img = MIMEImage(f.read())
            msg.attach(img)

            try:
                # Connect to the Gmail SMTP server
                with smtplib.SMTP_SSL('smtp.gmail.com', 465) as server:
                    # Log in to your Gmail account
                    server.login(SENDER_EMAIL, SENDER_PASSWORD)

                    # Send the email
                    server.sendmail(SENDER_EMAIL, RECEIVER_EMAIL, msg.as_string())

                print("Email sent successfully!")

            except Exception as e:
                print(f"An error occurred while sending the email: {e}")
        else:
            print("Cannot send email without GMAIL_ADDRESS and GMAIL_PASSWORD in Colab Secrets.")


    except Exception as e:
        print(f"An error occurred with the OpenAI API: {e}")
else:
    print("Cannot proceed without an OpenAI API key.")

```
