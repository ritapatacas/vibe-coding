### 👇 Setting things up

<font size="3" color="wheat">
✅ <strong>Step 1</strong>
</font><br>

> <font size="3" color="wheat">
**Enable the Google Sheets API**</font><br><ol><font size="3" color="darkkhaki">
  <li>Go to the <a href="https://cloud.google.com/cloud-console">Google Cloud Console.</a></li>
  <li>`Create a project` (right of "Google Cloud" logo).</li>
  <li>Search for `Google Sheets API`.</li>
  <li>Click it, then click `Enable`.</li>
  </font>
</ol>




> image.png

<font size="3" color="wheat">
✅ <strong>Step 2</strong>
</font><br>

> <font size="3" color="wheat">
**Create a Service Account**</font><br><ol><font size="3" color="darkkhaki">
  <li>Go to `IAM & Admin` → `Service Accounts`</li>
  <li>Click `Create Service Account`</li>
  <li>Give it a name (e.g., sheets-writer), click `Create` and `Continue`</li>
  <li>Skip assigning a role (you’ll give access by sharing the sheet)</li>
  <li>Click `Done`</li>
  </font>
</ol>

<font size="3" color="wheat">
✅ <strong>Step 3</strong>
</font><br>

> <font size="3" color="wheat">
**Create and Download a JSON Key**</font><br><ol><font size="3" color="darkkhaki">
  <li>In the list of service accounts, click the 3 dots next to your new account</li>
  <li>Click `Manage keys`</li>
  <li>Under `Keys`, click `Add Key` → `Create new key`</li>
  <li>Select `JSON`, click `Create`</li>
  <li>Save the .json file securely — you’ll use this in Python</li>
  </font>
</ol>



> <font size="3" color="wheat">*JSON (JavaScript Object Notation) is a type of file used for storing and exchanging data between applications*</font>



<font size="3" color="wheat">
✅ <strong>Step 4</strong>
</font><br>

> <font size="3" color="wheat">
**Share Your Sheet with the Service Account**</font><br><ol><font size="3" color="darkkhaki">
  <li>Open the Google Sheet you want to access</li>
  <li>Click `Share` in the top-right</li>
  <li>Add the service account email (something like sheets-writer@your-project.iam.gserviceaccount.com) with `Editor permission`.</li>
  <li>Click `Send`</li>
  </font>
</ol>
