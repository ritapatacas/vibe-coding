## <font size="5">5. &nbsp;&nbsp;Visualize data</font>

###
> 🔎&nbsp;&nbsp;<font size="3" color="lightsteelblue">Open your brand new `monthly_sales.csv` and use its data to make a graph</font>

####  <font size="2" color="gray">> code example</font>

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the CSV file
df_csv = pd.read_csv('monthly_sales.csv')

# Create a bar chart
plt.figure(figsize=(10, 6))  # Adjust figure size as needed
plt.bar(df_csv['Client Name'], df_csv['Monthly Sales'])
plt.xlabel('Client Name')
plt.ylabel('Monthly Sales')
plt.title('Monthly Sales per Client')
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

```python
from rembg import remove
from PIL import Image

# Load the image
input_image = Image.open("/content/vibecoding.jpg")

# Remove the background
output_image = remove(input_image)

# Save the image without background
output_image.save("/content/vibecoding_no_background.png")

print("Background removed and image saved as vibecoding_no_background.png")
```
